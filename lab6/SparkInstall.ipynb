# Pierwszy program

```python
from pyspark.sql import SparkSession
import findspark
def init_session():
	findspark.init()
	return SparkSession.builder.appName("Simple App").getOrCreate()

if __name__ == '__main__':
	session = init_session()
	print("Spark version: ",session.version)
```
